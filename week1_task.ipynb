```python
#Drop coloumns where more than 50% of the values are missing
threshold = len(df) * 0.5
df_clean = df.dropna(thresh=threshold, axis = 1)

print("Original coulumns:",df.shape[1])
print("Columns after dropping high missing ones:",df_clean.shape[1])
```

```python
#For core physical coloumns,fill missing with median values
key_cols = ['pl_orbper','pl_rade','pl_bmasse','st_teff','st_rad','st_mass']
key_cols = [c for c in key_cols if c in df_clean.columns]

for col in key_cols:
    median_value = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_value)

print("Missing values after filling with median:")
print(df_clean[key_cols].isnull().sum())
```

```python
#Drop rows with missing planets name and discovery year
df_clean = df_clean.dropna(subset=['pl_name', 'disc_year'])

#Ensure disc_year is integer
df_clean['disc_year'] = df_clean['disc_year'].astype(int)

print('final cleaned shape:', df_clean.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.hist(df_clean['disc_year'], bins=30,color='skyblue', edgecolor='black')
plt.xlabel('Discovery Year')
plt.ylabel('Number of planets Discovered')
plt.title('Distribution of Planetary Discoveries')
plt.show()
```
